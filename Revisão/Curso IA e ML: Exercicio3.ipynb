```python
notas = list(map(float, input("Digite 5 notas, separando cada uma por vírgula (Ex: 4,3,2,1,9): ").split(",")))
while len(notas) != 5:
  print("\nVocê não inseriu 5 notas. Digite novamente.\n")
  notas = list(map(float, input("Digite 5 notas, separando cada uma por vírgula (Ex: 4,3,2,1,9): ").split(",")))
```

```python
notas
```

```python
total = 0

for i in range(len(notas)):
  total += notas[i]

media = total / len(notas)

print(f"Total: {round(total,2)}",f"Média: {round(media,2)}")
```

```python
tabuada = int(input("Qual número você quer ver a tabuada? "))
final_tabuada = int(input("Você quer que a tabuada multiplique até que valor (Ex: se quer ver até 3x10, digite 10)? "))

```
