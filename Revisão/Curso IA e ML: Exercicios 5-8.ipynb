Crie uma classe chamada aluno com os seguintes atributos:
- Nome
- Nota 1
- Nota 2
- Crie um construtor para a classe (__init__)



Crie as seguintes funções (métodos):
- Calcula média, retornando a média aritmética entre as notas
- Mostra dados, que somente imprime o valor de todos os atributos
- Resultado, que verifica se o aluno está aprovado ou reprovado (se a média for maior ou igual a 6.0, o aluno está aprovado)



Crie dois objetos (aluno1 e aluno2) e teste as funções

Bom trabalho,

Jones

```python
class aluno:
  def __init__(self, nome, nota1, nota2):
    self.nome = nome
    self.nota1 = nota1
    self.nota2 = nota2
  def media(self):
    media = (self.nota1 + self.nota2)/2
    return media
  def imprimir(self):
    print(f'Nome do aluno: {self.nome}, Nota 1: {self.nota1}, Nota 2: {self.nota2}')
  def resultado(self):
    if (self.nota1 + self.nota2)/2 > 6: print("Aluno está aprovado!")
    else: print("Aluno está reprovado :(")
```

```python
aluno1 = aluno(nome='yasmin', nota1=10, nota2=9)

aluno1.media()
```

```python
aluno1.imprimir()
```

```python
aluno1.resultado()
```

Para revisar o conteúdo prático sobre expressões regulares, implemente o exercício abaixo. Logo em seguida você pode acessar a aula em vídeo com a solução

Crie expressões regulares para extrair as seguintes informações do texto abaixo (use a função findall):
- Números de telefone
- CEPs
- URLs

Bom trabalho,

Jones

```python
import re
```

```python
frase_bio = 'Olá, me chamo Alexandra e quero alterar o número de telefone (88)98227-6141 para (85)98877-6543. meu email é ale@email.com.br, o cep de casa é 60830-888 mas já morei no 61411-112. pode entrar em contato via linkedin https://linkedin.com.br ou github https://github.com'
```

```python
telefones = re.findall('\\(\\d{2}\\)\\d{5}-\\d{4}',frase_bio)
print('telefones encontrados: ', telefones)
```

```python
ceps = re.findall("6\\d{4}-\\d{3}", frase_bio)
print('CEPs encontrados: ', ceps)
```

```python
urls = re.findall("https://\\w+", frase_bio)
print('URLs encontrados: ', urls)
```

Considerando o dicionário com o nome dos alunos e suas respectivas notas abaixo, crie uma estrutura de repetição para percorrer cada elemento do dicionário para gravar cada aluno em um novo arquivo de texto
- Cada aluno deve ocupar uma linha do novo arquivo de texto
- O formato deve ser: nome,nota (Pedro,8.0)
- Após a criação do arquivo de texto, faça a leitura do arquivo e mostre todos os alunos

alunos = {'Pedro': 8.0, 'Maria': 10.0, 'Amilton': 7.5}
Bom trabalho,

Jones

```python
alunos = {'Pedro': 8.0, 'Maria': 10.0, 'Amilton': 7.5}
```

```python
with open ('texto2.txt','w') as arquivo_texto:
  for nome, notas in alunos.items():
    arquivo_texto.write(f'{nome},{notas}\n')

```

```python
with open ('texto2.txt','r') as arquivo_notas:
  for linha in arquivo_notas:
    print(linha)
```

Crie uma lista vazia e faça a leitura de dois valores do tipo float, colocando cada um dos valores nas primeiras posições da lista (o valor1 ficará na posição 0 da lista e o valor2 ficará na posição 1 da lista). Faça a divisão dos dois valores e trate as seguintes exceções:
- ValueError: se o usuário digitar um caracter
- ZeroDivisionError: se o usuário digitar zero e ocorrer erro na divisão
- IndexError: caso a divisão seja feita levando em consideração posições que não existem na lista
- KeyboardInterrupt: caso o usuário interrompa a execução



Mostre uma mensagem personalizada na ocorrência de cada um desses erros

Bom trabalho,

Jones

```python
while True:
  try:
    lista_inicial = list(map(float,input("Digite dois números float, separados por virgulas: ").split(",")))
  except KeyboardInterrupt:
     print("Você interrompeu a execução do código.")
     break
  except ValueError:
    print ("Você digitou algum caracter, tente novamente")
  else:
    print('Sua lista (float) é: ',lista_inicial)
    try:
      v0_v1= round(lista_inicial[0]/lista_inicial[1],2)
    except ZeroDivisionError:
      print("Erro na divisão pois o segundo valor é 0. Tente novamente.")
    else:
      print("A divisão do primeiro valor pelo segundo é: ", v0_v1)
      try:
        v1_v0= round(lista_inicial[1]/lista_inicial[0],2)
      except ZeroDivisionError:
        print("Erro na divisão pois o primeiro valor é 0. Tente novamente.")
      else:
        print("A divisão do segundo valor pelo segundo é: ", v1_v0)
        break
```
