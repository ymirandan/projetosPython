```python
try:
  numeros = list(map(int, input("Digite 5 números inteiros, separando cada um por vírgula (Ex: 4,3,2,1,9): ").split(",")))
except:
    print("\nVocê não inseriu 5 numeros inteiros. Rode o código novamente.\n")
while len(numeros) != 5 :
  print("\nVocê não inseriu 5 numeros inteiros. Digite novamente.\n")
  try:
    numeros = list(map(int, input("Digite 5 números inteiros, separando cada uma por vírgula (Ex: 4,3,2,1,9): ").split(",")))
  except:
    print("\nVocê não inseriu 5 numeros inteiros. Rode o código novamente.\n")
```

```python

soma = 0

for i in range(len(numeros)):
  soma += numeros[i]
print(f"Soma dos números digitados: {soma}")
```

```python
inicio_info = int(input("Você quer subir informações de quantos alunos? "))
dicionario_alunos = {}

for i in range(1,inicio_info +1):
  nome_nota = list(map(str, input("\nDigite o nome do aluno e sua nota, separando por vírgula (Ex: Marcos Paulo, 7.8) ").split(",")))
  dicionario_alunos.update({nome_nota[0]:float(nome_nota[1])})

novos_alunos = str(input("\nVocê quer subir informações de outro aluno? (Digite sim ou não) "))
if novos_alunos == "sim":
    nome_nota = list(map(str, input("\nDigite o nome do aluno e sua nota, separando por vírgula (Ex: Aline, 6.2) ").split(",")))
    dicionario_alunos.update({nome_nota[0]:float(nome_nota[1])})
    media = sum(dicionario_alunos.values())/len(dicionario_alunos.values())
    print(f"\nDicionário completo: {dicionario_alunos}\nMédia das notas: {round(media,2)}")
else:
    media = sum(dicionario_alunos.values())/len(dicionario_alunos.values())
    print(f"\nDicionário completo: {dicionario_alunos}\nMédia das notas: {round(media,2)}")
```

```python
matriz = np.array([[3, 4, 1], [3, 1, 5]])
```

```python
soma = 0

for i in range(matriz.shape[0]):
  for j in range(matriz.shape[1]):
    soma = soma + matriz[i][j]

print(soma)
```
